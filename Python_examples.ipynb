# Get sub modules of a module or package and its versions

```python
import pkgutil
import qdk # Example module

def list_submodules(package):
    """Lists the names of direct submodules within a package."""
    print(f"Submodules in {package.__name__}:")
    for importer, modname, ispkg in pkgutil.iter_modules(package.__path__):
        # modname will be just the name of the submodule, not the full dotted path
        print(f"* {modname} (is a package: {ispkg})")


list_submodules(qdk)
```

```python
# Get sub modules and the version if available
import pkgutil
import importlib.metadata
import importlib.util
import sys
import qdk # Example module

def get_module_version(module_name):
    """Attempts to find the version of a module."""
    try:
        # Try importlib.metadata first (modern approach)
        return importlib.metadata.version(module_name)
    except importlib.metadata.PackageNotFoundError:
        # Fallback to checking __version__ after importing (less reliable)
        try:
            spec = importlib.util.find_spec(module_name)
            if spec and spec.loader:
                module = importlib.util.module_from_spec(spec)
                sys.modules[module_name] = module
                spec.loader.exec_module(module)
                return getattr(module, '__version__', 'N/A (no __version__)')
        except ImportError:
            return "N/A (import error)"
    except Exception as e:
        return f"Error: {e}"

print("Found modules and their versions:")

def list_submodules(package):
    """Lists the names of direct submodules within a package."""
    print(f"Submodules in {package.__name__}:")
    for importer, modname, ispkg in pkgutil.iter_modules(package.__path__):
        # modname will be just the name of the submodule, not the full dotted path
        version = get_module_version(modname)
        #print(f"* {modname} (is a package: {ispkg})")
        print(f"* {modname} (is package: {ispkg}): Version: {version}")

list_submodules(qdk)
```
